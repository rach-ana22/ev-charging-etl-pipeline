```python
import sqlite3
```

```python
db_path = "..database/ev_charging.db"
```

```python
conn = sqlite3.connect(db_path)
```

```python
import pandas as pd

processed_path = "../data/processed"

charging_df = pd.read_csv(
    processed_path + "/charging_sessions.csv"
)

station_df = pd.read_csv(
    processed_path + "/station_summary.csv"
)
```

```python
charging_df.to_sql(
    "charging_sessions",
    conn,
    if_exists="replace",
    index=False
)

station_df.to_sql(
    "station_summary",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql(
    "SELECT COUNT(*) AS total_sessions FROM charging_sessions",
    conn
)
```

```python
pd.read_sql(
    "SELECT COUNT(*) AS total_stations FROM station_summary",
    conn
)
```

```python
#Top 5 stations by number of sessions

query = """
SELECT
    stationID,
    session_count,
    total_kWh
FROM station_summary
ORDER BY session_count DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
#Which hours have the highest charging demand?

query = """
SELECT
    hour,
    COUNT(*) AS session_count,
    SUM(kWhDelivered) AS total_kWh
FROM charging_sessions
GROUP BY hour
ORDER BY session_count DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
#top stations by total energy delivered.

query = """
SELECT
    stationID,
    session_count,
    total_kWh,
    avg_kWh_per_session
FROM station_summary
ORDER BY total_kWh DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```
